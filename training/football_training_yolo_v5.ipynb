```python
from dotenv import load_dotenv
import os

load_dotenv()
ROBOFLOW_API_KEY = os.getenv("ROBOFLOW_API_KEY")
if not ROBOFLOW_API_KEY:
    raise ValueError("ROBOFLOW_API_KEY is not set")
```

```python
from roboflow import Roboflow
rf = Roboflow(api_key=ROBOFLOW_API_KEY)
project = rf.workspace("roboflow-jvuqo").project("football-players-detection-3zvbc")
version = project.version(1)
dataset = version.download("yolov5")
                
```

```python
import shutil

# Check your exact folder name. If Roboflow named it slightly differently, change it here.
dataset_name = 'football-players-detection-1' 
nested_dir = os.path.join(dataset_name, dataset_name)

# 1. Create the nested directory if it doesn't exist
if not os.path.exists(nested_dir):
    os.makedirs(nested_dir)
    print(f"Created nested directory: {nested_dir}")

# 2. Safely move the folders
for split in ['train', 'test', 'valid']:
    src = os.path.join(dataset_name, split)
    dst = os.path.join(nested_dir, split)
    
    # If the source exists, move it
    if os.path.exists(src):
        shutil.move(src, dst)
        print(f"Successfully moved {src} to {dst}")
    # If the destination already exists, it means we already moved it!
    elif os.path.exists(dst):
        print(f"{split} is already in the nested folder. Skipping.")
    # If neither exists, the download might have failed
    else:
        print(f"WARNING: Could not find {src}. Did the Roboflow download finish?")
```
